#### Data cleaning

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates().reset_index(drop=True)

print("Rows after removing duplicates:", len(df))
```

#### Clean the price column

```python
df["price"] = df["price"].str.replace("₹", "", regex=False)
df["price"] = df["price"].str.replace(",", "", regex=False)
df["price"] = df["price"].str.strip()
df["price"] = pd.to_numeric(df["price"], errors="coerce")
df["price"].head()
```

#### Check invalid prices

```python
print("Invalid price values:", df["price"].isna().sum())
```

#### Clean km_driven

```python
df["km_driven"] = df["km_driven"].str.replace("km", "", regex=False)
df["km_driven"] = df["km_driven"].str.strip()

df["km_driven"] = pd.to_numeric(
    df["km_driven"],
    errors="coerce"
)

df["km_driven"].head()
```

```python
print("Invalid km values:", df["km_driven"].isna().sum())
```

```python
df[df["km_driven"].isna()]
```

```python
## as we got one invalid row means (null value) we can remove that row.
df = df.dropna(subset=["km_driven"]).reset_index(drop=True)

print("Rows after removing invalid km:", len(df))
```

```python
df.sort_values("price", ascending=False)[
    ["bike_name", "price", "km_driven", "ownership", "location"]
].head(10)
```

```python
# Check unusually high prices
df[df["price"] > 500000][
    ["bike_name", "price", "km_driven", "ownership", "location"]
].sort_values("price", ascending=False)
```
